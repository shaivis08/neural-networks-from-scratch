# Cross Entropy & Classification

- **Classification setup**
  - `X` → input features
  - `y` → class labels represented as integer class indices (`0, 1, 2, ...`)
  - Each sample belongs to one class.

- **Network output**
  - The final `Linear` layer produces **logits**.
  - For 3 classes, each sample produces 3 logits:
    - one score for class 0
    - one score for class 1
    - one score for class 2
  - Logits are raw scores, not probabilities.

- **Softmax**
  - Converts logits into probabilities.
  - Probabilities across classes sum to 1.
  - `ArgMax` can be used to select the class with the highest probability.
  - `CrossEntropyLoss` expects logits directly, so Softmax should **not** be applied before it.

- **Cross Entropy Loss**
  - Compares the model's predictions with the true class labels.
  - Low loss → high probability assigned to the correct class.
  - High loss → low probability assigned to the correct class.
  - Confidently wrong predictions receive a larger penalty.
  - PyTorch's `nn.CrossEntropyLoss()` takes:
    - raw logits
    - integer class labels

- **Training workflow**
  - `X → Linear → ReLU → Linear → logits`
  - `logits → CrossEntropyLoss → loss`
  - `loss.backward()` → calculates gradients
  - `optimizer.step()` → updates parameters
  - Repeat for multiple epochs.

- **Important distinction**
  - Before training, predictions are determined by randomly initialized parameters.
  - `y` provides the correct answers for the training examples.
  - Backpropagation and gradient descent adjust the weights so the network learns the relationship between `X` and `y`.

- **Experiment**
  - Trained a small 3-class classifier for 10 epochs.
  - Loss decreased from approximately `2.58 → 0.90`.
  - This shows that the model was learning from the labelled examples.

- **Key takeaway**
  - `Logits → Cross Entropy → Backpropagation → Parameter Update`
  - This is the basic training workflow for neural-network classification.

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
```

```python
X = torch.tensor([
    [1.0, 2.0, 3.0, 4.0],
    [2.0, 1.0, 4.0, 3.0],
    [5.0, 6.0, 7.0, 8.0],
    [6.0, 5.0, 8.0, 7.0],
    [9.0, 8.0, 7.0, 6.0],
    [8.0, 9.0, 6.0, 7.0]
])

y = torch.tensor([0, 0, 1, 1, 2, 2])
```

```python
class NN(nn.Module):
  def __init__(self):
    super().__init__()
    self.fc1 = nn.Linear(4,6)
    self.relu = nn.ReLU()
    self.fc2 = nn.Linear(6,3)
  def forward(self, x):
    x = self.fc1(x)
    x = self.relu(x)
    logits = self.fc2(x)
    return logits
```

```python
model = NN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)
epochs = 10
for epoch in range(epochs):
    logits = model(X)
    loss = criterion(logits, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```
