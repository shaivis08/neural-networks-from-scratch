**ReLU & Parameter Optimization**
* Added multiple layers to the neural network:
 * First Linear layer transforms the input features.
* ReLU introduces non-linearity.
Second Linear layer produces the final prediction.
* ReLU (Rectified Linear Unit):
* ReLU(x)=max(0,x)
* Negative values are converted to 0, while positive values remain unchanged.
Allows neural networks to learn non-linear relationships.
* During training, backpropagation calculates how each parameter contributed to the loss, and the optimizer uses these gradients to update the parameters.
* The decreasing loss shows that the model's parameters are being iteratively optimized to improve its predictions.
* Key takeaway: Adding ReLU between linear layers gives the network the ability to model non-linear relationships, while repeated gradient-based parameter updates progressively reduce the training loss.

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
X = torch.randn(6,5)
y = torch.randn(6,1)
```

```python
class NN(nn.Module):
  def __init__(self):
    super().__init__()
    self.fc1 = nn.Linear(5,4)
    self.relu = nn.ReLU()
    self.fc2 = nn.Linear(4,1)
  def forward(self, x):
    x = self.fc1(x)
    x = self.ReLU(x)
    x = self.fc2(x)
    return x
```

```python
model = NN()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)
epochs = 20
for epoch in range(epochs):
    predictions = model(X)
    loss = criterion(predictions, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```
