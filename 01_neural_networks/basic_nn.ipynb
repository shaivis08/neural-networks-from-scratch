**Notes — Basic Neural Network Training**

---


* A neural network is implemented as a class that inherits from nn.Module, which provides PyTorch's functionality for managing neural-network models and their parameters.
* The model's layers and parameters are defined in __init__().
The forward() function defines how the input passes through the layers of the network and produces an output.
* An object is instantiated from the neural-network class, a loss function is chosen to measure the difference between the model's predictions and the target values, and an optimizer is chosen to update the model's parameters.
* The optimizer applies gradient descent to update the weights and biases in order to minimize the loss.
* An epoch is one complete pass through the training data. Multiple epochs allow the model to repeatedly update its parameters and improve its predictions.
* optimizer.zero_grad() clears the gradients stored from the previous gradient calculation before calculating new ones.
* loss.backward() calculates the gradients of the loss with respect to the model's parameters.
* optimizer.step() updates the parameters using the calculated gradients. Conceptually:
  new_parameter = old_parameter - learning_rate × gradient





```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
X = torch.randn(4,3)
y = torch.randn(4,2)
```

```python
class SimpleNN(nn.Module):
    def __init__(self):
      super().__init__()
      self.layer = nn.Linear(3,2)
    def forward(self,x):
       out = self.layer(x)
       return out

model = SimpleNN()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)
epochs = 5
for epoch in range(epochs):
    predictions = model(X)
    loss = criterion(predictions, y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```

**Experiment observation**

---


* When only loss.backward() was used, the loss did not change because the gradients were calculated but the parameters were not updated.
* After adding optimizer.step(), the weights and biases changed and the loss decreased.
* Repeating this process over multiple epochs resulted in the loss decreasing further, demonstrating the basic training process of a neural network.
